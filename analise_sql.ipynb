```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("pedidos.db")

df = pd.read_csv("pedidos_limpos.csv")
df.to_sql("pedidos", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT COUNT(*) AS total_pedidos FROM pedidos", conn)
```

```python
pd.read_sql("SELECT * FROM pedidos LIMIT 5", conn)
```

```python
pd.read_sql("""
SELECT produto, quantidade, preco_unitario
FROM pedidos
WHERE categoria = 'Esporte'
  AND estado = 'RJ'
ORDER BY preco_unitario DESC
LIMIT 5
""", conn)
```

```python
pd.read_sql("""
SELECT produto, SUM(quantidade) AS unidades
FROM pedidos
GROUP BY produto
ORDER BY unidades DESC
""", conn)
```

```python
pd.read_sql("""
SELECT produto, SUM(receita) AS receita_total
FROM pedidos
GROUP BY produto
ORDER BY receita_total DESC
""", conn)
```

```python
pd.read_sql("""
SELECT categoria, SUM(receita) AS receita_total
FROM pedidos
GROUP BY categoria
ORDER BY receita_total DESC
""", conn)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS cancelados
FROM pedidos
WHERE status = 'Cancelado'
""", conn)
```

```python
pd.read_sql("""
SELECT ROUND(100.0 * SUM(CASE WHEN status = 'Cancelado' THEN 1 ELSE 0 END) / COUNT(*), 2) AS taxa_cancelamento
FROM pedidos
""", conn)
```

```python
pd.read_sql("""
SELECT categoria,
       ROUND(100.0 * SUM(CASE WHEN status = 'Cancelado' THEN 1 ELSE 0 END) / COUNT(*), 2) AS taxa_cancelamento
FROM pedidos
GROUP BY categoria
ORDER BY taxa_cancelamento DESC
""", conn)
```

```python
pd.read_sql("""
SELECT strftime('%Y-%m', data_pedido) AS mes, SUM(quantidade) AS unidades
FROM pedidos
GROUP BY mes
ORDER BY mes
""", conn)
```

```python
pd.read_sql("""
SELECT strftime('%Y-%m', data_pedido) AS mes, ROUND(SUM(receita), 2) AS receita_total
FROM pedidos
GROUP BY mes
ORDER BY mes
""", conn)
```

```python
pd.read_sql("""
WITH vendas AS (
    SELECT estado, produto, SUM(quantidade) AS unidades
    FROM pedidos
    GROUP BY estado, produto
),
ranking AS (
    SELECT estado, produto, unidades,
           ROW_NUMBER() OVER (PARTITION BY estado ORDER BY unidades DESC, produto) AS posicao
    FROM vendas
)
SELECT estado, produto, unidades
FROM ranking
WHERE posicao = 1
ORDER BY estado
""", conn)
```
